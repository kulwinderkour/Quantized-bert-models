```python
def calculate_scale_and_zeropoint(min_val,max_val,num_bits=8):   #this functions store maximum and minimum value
    qmin,qmax=0,(2**num_bits)-1
    min_val = min(min_val,0)
    max_val = max(max_val,0)
    if(min_val == max_val):
        return 1.0,0
    #S = (max - min) / (qmax - qmin)
    scale = (max_val-min_val)/(qmax - qmin)
    inital_zero_point = qmin - (min_val /scale)
    zero_point = int(np.clip(np.round(inital_zero_point),qmax,qmin))
    return scale,zero_point 
```

```python
# this is quantization function
def quanitze(tensor,scale,zero_point,num_bits=0):
    qmin,qmax = 0,(2**num_bits) -1
    #q = round(x/s) + zero_point
    q_tensor = np.round(tensor/scale)+ zero_point
    return np.clip(q_tensor,qmin,qmax).astype(np.unit8)
```

```python
def dequantize(q_tensor,scale,zero_point):
    # q2 = s(q-z)
    
    return scale*(q_tensor.astype(np.float32) - zero_point)
```
